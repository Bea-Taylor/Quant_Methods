```python
import pandas as pd
import numpy as np
from pathlib import Path
import statsmodels.api as sm

# little function to define the file root on different machines
def find_qm_root(start_path: Path = Path.cwd(), anchor: str = "_quarto.yml") -> Path:
    """
    Traverse up from the start_path until the anchor folder (e.g. 'QM' or 'QM_Fork')      is found. Returns the path to the anchor folder.
    """
    for parent in [start_path] + list(start_path.parents):
        if (parent / anchor).exists() or parent.name == anchor:
            return parent
    raise FileNotFoundError(f"Anchor folder '{anchor}' not found in path      hierarchy.")
  
qm_root = find_qm_root()

print(f"""----------------------------------------
QM root found at: {qm_root}
Moving on to data processing...
----------------------------------------""")


base_path = qm_root / "practicals" / "data" / "Performancetables_130242" / "2022-2023"
na_all = ["", "NA", "SUPP", "NP", "NE", "SP", "SN", "LOWCOV", "NEW", "SUPPMAT", "NaN"]

england_filtered = pd.read_csv(base_path / "england_filtered.csv", na_values=na_all, dtype={"URN": str})

# Log-transform safely: replace non-positive values with NaN
england_filtered['log_ATT8SCR'] = np.where(england_filtered['ATT8SCR'] > 0, np.log(england_filtered['ATT8SCR']), np.nan)
england_filtered['log_PTFSM6CLA1A'] = np.where(england_filtered['PTFSM6CLA1A'] > 0, np.log(england_filtered['PTFSM6CLA1A']), np.nan)

# Drop rows with NaNs in either column
england_filtered_clean = england_filtered.dropna(subset=['log_ATT8SCR', 'log_PTFSM6CLA1A'])

print(f""" ----------------------------------------
Data processing complete. 
Moving on to regression analysis... 
----------------------------------------""")


# Define independent and dependent variables
X = sm.add_constant(england_filtered_clean['log_PTFSM6CLA1A'])  # adds intercept
y = england_filtered_clean['log_ATT8SCR']

# Fit the model
england_model1 = sm.OLS(y, X).fit()
#england_summary = extract_model_summary(england_model1, 'England Model')

# Print summary
print(england_model1.summary())
```

### Take a moment to look at the output table. 
Before we move on to a multi-variable analysis, Can you understand the results of our basic model? Ask yourself:

2. What is our independent variable?
3. How many observations do we have, and how many degrees of freedom? Are they enough?
4. What is the coefficient of our independent variable? What does it mean given that our variables went through log-transformation?
5. What does the R-squared value tell us about how much of the variation in grades this single factor explains?
6. Is it statistically significant?
7. What is our 95% CI (Confidence Interval?) how does it relate to our coefficient?
8. What does the t-value tell us?

### Answers

1. log_ATT8SCR - log transformed attainment 8 scores of schools
2. log_PTFSM6CLA1A - log transformed percentage of disadvantaged students
3. 2968 observations, and the degrees of freedom is 2966, more than enough. A common rule of thumb is the 10:1 ratio - 10 observations for every 1 parameter - although this is a rough guide. If in doubt, even up to 20:1 to be safe. We have 2966 DFs for 2 parameters!
4. Negative -0.2, meaning that for every 1% increase in percentage of disadvantaged students, the expected attainment 8 score drops in around .2%.
5. With an R-squared of 0.468, we can assume that the model explains 47% of the variation in Attainment 8 scores.
6. Yes, the model is statistically significant, as p_value is smaller than 0.05.
7. The 95 CI of our model is between -0.215 to -0.199, (roughly two std errors from either sides of our coefficient)
8. t-values can be thought of as a standardised coefficient - very useful for comparing the relative importance of different predictors in the model. The larger the t-value, the more important the predictor is in explaining the variation in the dependent variable - more important the variable.

### Changing your dummy reference category

Your dummy variable reference category is the category within the
variable that all other categories will be compared against in your
model.

While the reference category has no effect on the model itself, it does
make a difference for how you interpret your model.

For example, if you are using Regions in England as a dummy, setting
your reference region as London will mean all other regions are compared
to it and they might naturally be lower or higher.

A good strategy is to select the most or least numerous or important
category in your variable, rather than something in the middle. Of
course, you may not know which is most important until you run your
model, so you may need to go back and reset the reference variable and
run the model again.

Here is some code in R and Python to help you carry out the setting of
the reference level (Yes, it’s more straightforward in R!)


```python
import pandas as pd
from pandas.api.types import CategoricalDtype

# Ensure it's a copy to avoid SettingWithCopyWarning
england_filtered_clean = england_filtered_clean.copy()

"""
The next bits of code take the desired reference (baseline) category and place it as the first category in each categorical variable.
This is important for regression analysis, as the first category becomes the benchmark against which all other categories are compared.

The code works as follows:
1. For each categorical variable (gor_name, ofsted_rating_name, ADMPOL_PT), we define a new CategoricalDtype with the desired reference category listed first (in this example, South East for gor_name, for example).
2. We use set() to collect unique values without duplicates, remove the reference category, sort the rest alphabetically, and prepend the reference category to the front
    For example, for ofsted_rating_name, we set "Good" as the reference category and sort the rest of the categories alphabetically: "Good, Outstanding, Requires Improvement, Serious Weaknesses, Special Measures"
3. We then convert each column to this new categorical type so the reference level is locked in when dummy variables are created for the regression.
"""


gor_dtype = CategoricalDtype(
    categories=["South East"] + sorted(set(england_filtered_clean["gor_name"]) - {"South East"}),
    ordered=True
)


england_filtered_clean["gor_name"] = england_filtered_clean["gor_name"].astype(gor_dtype)

# Set reference level for ofsted_rating_name
ofsted_dtype = CategoricalDtype(
    categories=["Good"] + sorted(set(england_filtered_clean["OFSTEDRATING"].dropna()) - {"Good"}),
    ordered=True
)
england_filtered_clean["ofsted_rating_name"] = england_filtered_clean["OFSTEDRATING"].astype(ofsted_dtype)

# Set reference level for ADMPOL_PT
admpol_dtype = CategoricalDtype(
    categories=["OTHER NON SEL"] + sorted(set(england_filtered_clean["ADMPOL_PT"]) - {"OTHER NON SEL"}),
    ordered=True
)
england_filtered_clean["ADMPOL_PT"] = england_filtered_clean["ADMPOL_PT"].astype(admpol_dtype)
```

### Reclassifying a variable into fewer categories

As mentioned in the lecture, it can sometimes be useful to reclassify a
variable into fewer categories to see if a signal appears.

Let’s have a look at the religious character of a school variable and
test it out as dummy in a basic model:

Note the Python code is far more complicated than the R code, but should
produce similar outputs

-- Note that this example involves a different variable than the previous three we prepared for regression!

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Drop rows with missing values in relevant columns
model_df = england_filtered_clean[['ATT8SCR', 'PTFSM6CLA1A', 'RELCHAR']].dropna().copy()

# Log-transform numeric variables
model_df['log_ATT8SCR'] = np.log(model_df['ATT8SCR'].astype(float))
model_df['log_PTFSM6CLA1A'] = np.log(model_df['PTFSM6CLA1A'].astype(float))

# Ensure RELCHAR is treated as categorical
model_df['RELCHAR'] = model_df['RELCHAR'].astype('category')

# Create design matrix with dummy variables for RELCHAR
X = pd.get_dummies(model_df[['log_PTFSM6CLA1A', 'RELCHAR']], drop_first=True)

# Add constant
X = sm.add_constant(X)

# Response variable
y = model_df['log_ATT8SCR']

# Ensure all columns are float64 to avoid dtype issues
X = X.astype('float64')
y = y.astype('float64')

# Fit linear model using Pandas DataFrame directly
model = sm.OLS(y, X).fit()

# Display summary with proper variable names
print(model.summary())
```

You will notice that the religious character variables nearly all
insignificant. As such, lets try collapsing into three groups - “None”,
“Christian” and “Non-Christian”


```python
# Define mapping function
def classify_relchar(value):
    if pd.isna(value):
        return "None"
    elif value in ["Does not apply", "None"]:
        return "None"
    elif value in [
        "Roman Catholic", "Greek Orthodox", "Church of England", "Catholic", "Christian",
        "Church of England/Roman Catholic", "Roman Catholic/Church of England",
        "Church of England/Christian", "Anglican/Church of England"
    ]:
        return "Christian"
    else:
        return "Non-Christian"

# Apply classification
england_filtered_clean['RELCHAR_Grp'] = england_filtered_clean['RELCHAR'].apply(classify_relchar)
```

Now re-run your model


```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Convert RELCHAR_Grp to a categorical variable with "None" as the reference level
england_filtered_clean['RELCHAR_Grp'] = pd.Categorical(
    england_filtered_clean['RELCHAR_Grp'],
    categories=["None", "Christian", "Non-Christian"],
    ordered=False
)

# Drop rows with missing values in relevant columns
model_df = england_filtered_clean[['ATT8SCR', 'PTFSM6CLA1A', 'RELCHAR_Grp']].dropna().copy()

# Log-transform numeric variables
model_df['log_ATT8SCR'] = np.log(model_df['ATT8SCR'].astype(float))
model_df['log_PTFSM6CLA1A'] = np.log(model_df['PTFSM6CLA1A'].astype(float))

# Ensure RELCHAR is treated as categorical
model_df['RELCHAR_Grp'] = model_df['RELCHAR_Grp'].astype('category')

# Create design matrix with dummy variables for RELCHAR
X = pd.get_dummies(model_df[['log_PTFSM6CLA1A', 'RELCHAR_Grp']], drop_first=True)

# Add constant
X = sm.add_constant(X)

# Response variable
y = model_df['log_ATT8SCR']

# Ensure all columns are float64 to avoid dtype issues
X = X.astype('float64')
y = y.astype('float64')

# Fit linear model using Pandas DataFrame directly
model = sm.OLS(y, X).fit()

# Display summary with proper variable names
print(model.summary())
```

Has this made a difference? What’s happened to the significance of the
religious character variable?

## Task 4b - Which variable *really* matters most?

The first question in Task 4 asked you to check the t-values. That is
the usual answer, and it is shakier than it looks - as the lecture
showed, a t-value is $\frac{\text{Estimate}}{\text{SE}}$, so it mixes
**how big** an effect is with **how precisely** you have measured it.
With a few thousand schools, everything is precise.

The deeper problem is that your predictors **overlap**. Disadvantage and
absence share a great deal of their explanatory power, and the model has
to split that shared portion between them somehow.

### How much is actually shared?

Fit each predictor on its own and compare with the full model:

```python
def r2_of(vs):
    rhs = " + ".join(vs) if vs else "1"
    return smf.ols(f"np.log(ATT8SCR) ~ {rhs}", data=england_filtered_clean).fit().rsquared

r2_of(["np.log(PTFSM6CLA1A)"])
print(f"R-squared for model with log(PTFSM6CLA1A): {r2_of(['np.log(PTFSM6CLA1A)'])}")
r2_of(["np.log(PERCTOT)"])
print(f"R-squared for model with log(PERCTOT): {r2_of(['np.log(PERCTOT)'])}")
r2_of(["np.log(PTFSM6CLA1A)", "np.log(PERCTOT)"])
print(f"R-squared for model with log(PTFSM6CLA1A) and log(PERCTOT): {r2_of(['np.log(PTFSM6CLA1A)', 'np.log(PERCTOT)'])}")
```

> **Do the arithmetic before reading on**
>
> Add the two solo $R^2$ values together, then compare with the $R^2$ of
> the model containing both. The sum will come out **well above** the
> joint figure. The difference is variance the two predictors
> **share** - explained by either one of them, owned by neither.

### Splitting the shared variance fairly - the LMG / Shapley decomposition

Rather than letting the model carve up the shared portion arbitrarily,
we can **average each variable’s contribution over every possible order
of entry**. Add disadvantage first then absence; then absence first then
disadvantage; average what each gained. That average is the **Shapley
value**, known in regression as the **LMG decomposition**.

The function below does it for any number of predictors. There is a
package for this - `relaimpo::calc.relimp(model, type = "lmg")` - but it
is not installed here, and writing it out shows you exactly what is
being averaged.


```python
from itertools import combinations
from math import factorial

def lmg_shares(outcome, predictors, data):
    def r2(vs):
        rhs = " + ".join(vs) if vs else "1"
        return smf.ols(f"{outcome} ~ {rhs}", data=data).fit().rsquared
    k = len(predictors)
    phi = {}
    for v in predictors:
        others = [p for p in predictors if p != v]
        total = 0.0
        for size in range(k):
            w = factorial(size) * factorial(k - size - 1) / factorial(k)
            for S in combinations(others, size):
                total += w * (r2(list(S) + [v]) - r2(list(S)))
        phi[v] = total
    s = sum(phi.values())
    return pd.DataFrame({"variable": list(phi), "lmg": list(phi.values()),
                         "share": [x / s for x in phi.values()]})

lmg_shares("np.log(ATT8SCR)",
           ["np.log(PTFSM6CLA1A)", "np.log(PERCTOT)", "np.log(PNUMEAL)"],
           england_filtered_clean)
```

> **What you should get**
>
> On the three-predictor model above, both languages give the same
> answer:
>
> | Variable          | $R^2$ alone |   LMG | Share of explained |
> |-------------------|------------:|------:|-------------------:|
> | % Disadvantaged   |       0.469 | 0.301 |          **41.6%** |
> | % Overall Absence |       0.610 | 0.401 |          **55.5%** |
> | % EAL             |       0.009 | 0.021 |           **3.0%** |
>
> The LMG values **sum to 0.724**, which is exactly the $R^2$ of the
> full three-predictor model. Nothing is lost and nothing is counted
> twice - that is the property that makes the decomposition worth
> having.
